# Data Types

```python
# Data Types
a = 10 # integer: numbers without decimal point
print(a)
b = 10.0 # float: numbers with decimal point
print(b)
c = '10' # string: characters enclosed in single or double quotes
print(c)
d = True # boolean: True or False
print(d)
e = [1, 2, 3] # list: collection of items enclosed in square brackets
print(e)
f = (1, 2, 3) # tuple: collection of items enclosed in parentheses
print(f)
g = {'a': 1, 'b': 2, 'c': 3} # dictionary: collection of key-value pairs enclosed in curly brackets
print(g)
h = {1, 2, 3} # set: collection of unique items enclosed in curly brackets
print(h)
i = None # NoneType: represents the absence of a value
print(i)
j = "" # str-Blank "" is a string value with no characters inside; not same as None
print(j)
k = " " # str-White Space " " is a string value with a single or more white spaces character inside; not same as None
print(k)


```
