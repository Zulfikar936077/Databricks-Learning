## 1.1 type()

### _Example 1: Show the types of strings._

```python
name = "Hasan"; # string
print(type(name));
age = 25; # integer
print(type(age));
height = 1.75; # float
print(type(height));
is_married = False; # boolean
print(type(is_married));
print(name, age, height, is_married);
```

## 1.2 str()

### _Example 2: Convert an integer data type to string and show it in a sentence with another string variable._

```python
# If we want to combine str data type with integer as follows, it will show error:
# print( name + " is " + age + " years old.");
# That's why we need to convert integer to string
print( name + " is " + str(age) + " years old.");


print(type(age)); # the data type remains integer
# print(age = age + 1); # that means the int variable will only be added with another int not str
age = str(age); # to permanenently change the data type from int to str
print(type(age)); # checking the change of data type
```

## 1.3 len()

### _Example 3: Show with an example if the lenth of a password is long enough or shorter than required length._

```python
password = "1234"
print(len(password));
if len(password) < 8:
  print("Password is too short");
else:
  print("Password is long enough");

password2 ="  123456"
print(len(password2));
if len(password2) < 8:
    print("Password is too short");
else:
    print("Password is long enough")
```
