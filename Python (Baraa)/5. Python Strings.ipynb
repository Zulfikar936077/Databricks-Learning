## 3.3. f-string

### _Example 8: Show the use cases of f-string._

```python
name = "Michael Jordan";
age = 55;
isStudent = False;
# the primary way of concatenating different data types
print(name + " is " + str(age) + " year old and his student status is " + str(isStudent));

# Modern way using f-string
# Use case-1 (for variables)
print(f"{name} is {age} year old and his student status is {isStudent}"); 
# Use case-2 (for expressions)
print(f"2 + 3 = {2+3}"); 
# Use case-3 (to include the {} in the output)
print(f"{{Can you see the curly braces?}}");


```

## 3.6 Indexing & Slicing

```python
text = "Python"
date = "2025-09-20"
# extract the first character
print(text[0]);
print(text[-6]);
# extract the last character
print(text[5]);
print(text[-1]);
# extract h 
print(text[3]);
print(text[-3]);

# extract the year
print(date[0:4]); # use the positive number when you have to extract s/th from the left
print(date[:4]);
print(date[-11:-6]); # use the positive number when you have to extract s/th from the right

# extract the month
print(date[5:7]);
print(date[-5:-3]);

# extract the day
print(date[8:]);
print(date[-2:]);
```

# 4. Cleaning

### _Example 11: Show the use cases of lstrip(), rstrip(), and strip()._

```python
# removing white space from left 
text1 = " Engineering"
print(text1.lstrip())

# removing white space from right
text2 = "Engineering "
print(text2.rstrip())

# removing white space from both sides
text3 = " Engineering "
print(text3.strip())

# strip does not remove white between the words
text4 = " Engineering is fun "
print(text4.strip())

# removing special characters
text5 = "###Abc###"
print(text5.strip("#"))
```

### _Example 12: Check whether the data is cleaned or is there any unwanted white space?_

### _Example 14: Show the use cases of upper() and lower() cases._

```python
# Use of cases 
text7 = " data Engineering"
print(text7.upper()) # Uppercase
print(text7.lower()) # Lowercase
print(text7.title()) # Titlecase

# Case conversion for matching the value
search = " Email"
data = " email "
# To check whether the values match or not
print(search == data) 
# To check whether the values match or not after converting
print(search.lower() == data.lower())
# To check whether the values match or not after converting and removing unwanted white space
# Best practice: to trim the value before search
print(search.lower().strip() == data.lower().strip()) 
```

### _Example 15: Turn the following messy string into a single clean summary with name, role, and age:_
> "968-Maria, (D@t@ Engineer ) ;; 27y "

The expected string is:
> name: maria | role: data engineer | age: 27

```python
string = "968-Maria, (D@t@ Engineer ) ;; 27y  "
name = "name: " + string[4:9].lower()
role = " | role: " + string[11:27].lower().replace("@", "a").replace("(", "").replace(")", "")
age = "| age: " + string[30:33]
finaleString = name + role + age
print(finaleString)
```

# 5. Search

### _Example 16: Show all the use cases of startswith(), endswith(), in and find()._

```python
# Is the phone number from Germany?
phone = "+49-176-12345"
phone2 = "49-176-12345"
phone3 = "0049-176-12345"
print(phone.startswith("+49"))

# Is the email address of Google?
email = "maria@google.com"
print(email.endswith("google.com"))

# Is the email address valid?
print("@" in email)

# Check if the url is an api endpoint
url = "https://www.google.com/api/search"
print("api" in url)

# Retrieve only the phone number without country code
print(phone[phone.find("-")+1:])
print(phone2[phone2.find("-")+1:])
print(phone3[phone3.find("-")+1:])
```

# 6. Validation

### _Example 17: Show all the use cases of isalpha() and is numeric()._

```python
# Use if isalpha() to check whether only letters exist in the string value
country = "USA"
print(country.isalpha()) # Country is valid
country1 = "USA1"
print(country1.isalpha()) # Country is not valid because there is a numeric value

# Use of isnumeric() check whether only digits exist in the string value
phone = "123456789"
print(phone.isnumeric()) # Phone number is valid
phone1 = "123456789a"
print(phone1.isnumeric()) # Phone number is not valid because there is a letter

```
