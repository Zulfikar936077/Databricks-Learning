# 1. STRING FUNCTIONS

### _Example 1: Show a list of customers' first names together with their country in one column._

```sql
%sql
SELECT 
    FirstName,
    Country,
    CONCAT(FirstName, ' ', Country) AS customer_info
FROM workspace.sql_course_db.customers
```

### _Example 2: transform the customer's first name to a lowercase._

```sql
%sql
SELECT firstname,
lower(firstname)
FROM workspace.sql_course_db.customers
```

### _Example 3: transform the customer's first name to a uppercase._

```sql
%sql
SELECT firstname,
upper(firstname)
FROM workspace.sql_course_db.customers
```

### _Example 4: Find customers whose first name contains leading or trailing spaces._

```sql
%sql
SELECT
    firstname
FROM 
    workspace.sql_course_db.customers
WHERE 
    FirstName != trim(FirstName)
```

### _Example 5: Replace dashes (-) from employees' Birth Date with slashes (/)_

```sql
%sql
SELECT 
    FirstName,
    BirthDate,
    replace(BirthDate, '-', '/') AS cleaned_birthday
FROM workspace.sql_course_db.employees
```

### _Example 6: Replace dashes (-) from a phone number_

```sql
%sql
SELECT
    '123-456-789' AS phone_number,
    replace('123-456-789', '-', '/') AS cleaned_phone_number
```

### _Example 7: Replace FIle Extension from .txt to .csv_

```sql
%sql
SELECT
    'Report.txt' AS old_filename,
    REPLACE('Report.txt', '.txt', '.csv') AS new_filename
```

### _Example 8: Calculate the length of each customer's first name._

```sql
SELECT
    firstname,
    len(firstname) AS length_firstname
FROM workspace.sql_course_db.customers
```

### _Example 9: Retrieve the first two characters of each first name._

```sql
SELECT
    firstname,
    LEFT(FirstName, 2) AS first_two_characters
FROM workspace.sql_course_db.customers
```

### _Example 10: Retrieve the last two characters of each first name._

```sql
SELECT
    firstname,
    RIGHT(firstname, 2) last_two_characters
FROM workspace.sql_course_db.customers
```

### _Example 11: After second character extract two characters._

```sql
SELECT
    firstname,
    SUBSTRING(firstname, 3, 2) AS extracted_characters
FROM workspace.sql_course_db.customers
```

### _Example 12: After second character extract all characters of the first name._

```sql
SELECT
    firstname,
    substring(firstname, 3, length(firstname)) AS remaining_characters
FROM workspace.sql_course_db.customers
```

### _Example 13: Retrieve a list of customers' first names after removing the first character._

```sql
SELECT
    firstname,
    substring(trim(firstname), 2, length(firstname)) AS sub_name
FROM workspace.sql_course_db.customers
```

# 2. NUMERIC FUNCTIONS

### _Example 14: Round 3.516 to 0, 1, and 2 positions._

```sql
SELECT 3.516,
ROUND(3.516, 2) AS ROUND_2,
ROUND(3.516, 1) AS ROUND_1,
ROUND(3.516, 0) AS ROUND_0
```

### _Example 15: Convert the -10 into a +10._

```sql
SELECT -10,
ABS(-10) AS Absolute_10
```

# 3. DATE & TIME

### _Example 16: Retrieve the Order Date, Ship Dat, and Creation Date of each Order from orders table._

```sql
SELECT
    OrderID,
    OrderDate,
    ShipDate,
    CreationTime
FROM
    workspace.sql_course_db.orders
```

### _Example 17: Retrieve the Order Date, Ship Dat, and Creation Date of each Order from orders table and also create a Hard Coded Column for reference._

```sql
SELECT
    OrderID,
    OrderDate,
    ShipDate,
    CreationTime,
    '2026' AS Current_Year
FROM
    workspace.sql_course_db.orders
```

### _Example 18: Retrieve the Order Date, Ship Dat, Creation Date, and Ref. Year of each Order from orders table. Then add the current date column for each order._

```sql
SELECT
    OrderID,
    OrderDate,
    ShipDate,
    CreationTime,
    '2026' AS Current_Year,
    getdate() AS Current_Date
FROM
    workspace.sql_course_db.orders
```
