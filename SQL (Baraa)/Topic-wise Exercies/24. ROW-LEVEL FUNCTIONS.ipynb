### _Example 3: transform the customer's first name to a uppercase._

```sql
%sql
SELECT firstname,
upper(firstname)
FROM workspace.sql_course_db.customers
```

### _Example 4: Find customers whose first name contains leading or trailing spaces._

```sql
%sql
SELECT
    firstname
FROM 
    workspace.sql_course_db.customers
WHERE 
    FirstName != trim(FirstName)
```

## 3.1 Part Extraction

### _Example 19: Retrieve the day, month, and year of  each Creation Date of each order._

```sql
SELECT
    CreationTime,
    year(CreationTime) AS Year,
    month(CreationTime) AS Month,
    day(CreationTime) AS Day
FROM
    workspace.sql_course_db.orders
```

### _Example 20: Retrieve the quarter, week, and hour of  each Creation Date of each order._

```sql
-- DATEPART() is used for other parameters (e.g., quarter, week, hour) not for Year, Month, and Day as these three have their own functions.
SELECT
    CreationTime,
    year(CreationTime) AS Year,
    month(CreationTime) AS Month,
    day(CreationTime) AS Day,
    date_part('quarter', CreationTime) AS Quarter_dp, -- Quarter
    date_part('week', CreationTime) AS Week_dp,
    date_part('hour', CreationTime) AS Hour_dp
FROM
    workspace.sql_course_db.orders
```

### _Example 21: Retrieve the name of each month and day of each Creation Date of each order._

```sql
-- DATENAME() is used to name the Month or Day instead of using numbers. But, in Databricks the DATENAME() is invalid as it is a T-SQL synthax. Here, the equivalent syntax is date_format().
SELECT
    CreationTime,
    month(CreationTime) AS Month,
    date_format(CreationTime, 'MMMM') AS Month_dn,
    day(CreationTime) AS Day,
    date_format(CreationTime, 'EEEE') AS DayName_dn, -- Get the name of each day
    date_part('quarter', CreationTime) AS Quarter_dp,
    date_part('week', CreationTime) AS Week_dp,
    date_part('hour', CreationTime) AS Hour_dp
FROM
    workspace.sql_course_db.orders
```

### _Example 22: Use DATETRUNC function for each Creation Date of each order._

```sql
--date_trun()
SELECT
    CreationTime,
    date_trunc('minute', CreationTime) AS Minute_dt,
    date_trunc('hour', CreationTime) AS Hour_dt,
    date_trunc('day', CreationTime) AS Day_dt,
    date_trunc('month', CreationTime) AS Month_dt,
    date_trunc('year', CreationTime) AS Year_dt,
    -- DATENAME()
    month(CreationTime) AS Month,
    date_format(CreationTime, 'MMMM') AS Month_dn,
    day(CreationTime) AS Day,
    date_format(CreationTime, 'EEEE') AS DayName_dn, -- Get the name of each day
    -- DATEPART()
    date_part('quarter', CreationTime) AS Quarter_dp,
    date_part('week', CreationTime) AS Week_dp,
    date_part('hour', CreationTime) AS Hour_dp
FROM
    workspace.sql_course_db.orders
```

### _Example 23: Retrieve the number of orders per month based on the Creation Date of the orders._

```sql
SELECT  
    date_trunc('month', CreationTime) AS month_start,
    count(*) AS NrOfOrders,
    date_format(date_trunc('month', CreationTime), 'MMMM') AS Month
FROM 
    workspace.sql_course_db.orders
GROUP BY 
    date_trunc('month', CreationTime)
```

### _Example 24: Retrieve the last day and first day of each month based on the Creation Date of the orders._

```sql
-- EOMONTH() is a T-SQL syntax. In databricks, the similar syntax is LAST_DAY().
SELECT  
   OrderID,
   CreationTime,
   LAST_DAY(CreationTime) AS month_end,
   CAST(date_trunc('month', CreationTime) AS DATE) AS month_start,
   DAY(LAST_DAY(CreationTime)) AS Last_Day,
   DAY(CAST(date_trunc('month', CreationTime) AS DATE)) AS First_Day
FROM 
    workspace.sql_course_db.orders
```

### _Example 25: How many orders were placed each year?_

```sql
SELECT
    YEAR(OrderDate) AS Year,
    count(*) AS NrOfOrders
FROM
    workspace.sql_course_db.orders
GROUP BY
   YEAR(OrderDate)
```

### _Example 26: How many orders were placed each year?_

```sql
SELECT 
    date_format(OrderDate, 'MMMM') AS Month_dn,
    count(*) AS NrOfOrders
FROM 
    workspace.sql_course_db.orders
GROUP BY 
   date_format(OrderDate, 'MMMM')
```

### _Example 27: Show all orders that were placed during the month of February?_

```sql
SELECT
    date_format(OrderDate, 'MMMM') AS Month_dn,
    count(*) AS NrOfOrders
FROM 
    workspace.sql_course_db.orders
WHERE
    month(OrderDate) = 2
-- Here, I used month(OrderDate) = 2 instead of  date_format(OrderDate, 'MMMM') = 'February' because integer is always faster than strings
GROUP BY
    date_format(OrderDate, 'MMMM')
```

## 3.2 Format & Casting

### _Example 28: Retrieve creaction date of each order and manipulate them into every FORMAT._

```sql
SELECT 
    OrderID,
    CreationTime AS INT_FORMAT,
    date_format(CreationTime, 'MM-dd-yyyy') AS USA_FORMAT,
    date_format(CreationTime, 'dd/MM/yyyy') AS EU_FORMAT,
    month(CreationTime) AS MonthNum, -- Here, month() cannot used for UK_Format because the value is a string
    date_format(CreationTime, 'dd') AS NrOfDays, -- Day of the Week as number
    date_format(CreationTime, 'EE') AS DayOfWeek, -- 'EE' gives short day name instead of 'ddd' in T-SQL
    date_format(CreationTime, 'EEEE') AS DayOfWeek, -- 'EEEE' gives full day name 
    date_format(CreationTime, 'MM') AS Month, -- Month as number
    date_format(CreationTime, 'MMM') AS MonthName, -- 'MMM' gives short month name and 'MMMM' gives full month name 
    date_format(CreationTime, 'MMMM') AS MonthName 

FROM workspace.sql_course_db.orders
```

### _Example 29: Show CreationTime using the following format:_
        Day Wed Jan Q1 2025 12:345:56 PM

```sql
SELECT
    OrderID,
    CreationTime AS INT_FORMAT,
    CONCAT('Day ', date_format(CreationTime, 'EEE'), ' ', date_format(CreationTime, 'MMM'), ' ', 'Q', date_part('quarter', CreationTime), ' ', date_format(CreationTime, 'yyyy'), ' ', date_format(CreationTime, 'hh:mm:ss a')) AS CustomFormat
-- a is used to ensure am/pm is displayed
FROM    
    workspace.sql_course_db.orders
```

### _Example 30: Show the number of orders per month of 2025._

```sql
SELECT
    date_format(OrderDate, 'MMM yy') AS OrderDate,
    count(*)
FROM
    workspace.sql_course_db.orders
GROUP BY
    date_format(OrderDate, 'MMM yy')
```
