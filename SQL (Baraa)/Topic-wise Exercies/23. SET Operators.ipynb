# SET OPERATORS

## 1. UNION OPERATPOR

### _Example 1: Combine the data from employees and customers into one table._

#### Step-1: Query of both employees and customers tables

```sql
SELECT * FROM workspace.sql_course_db.employees
```

```sql
SELECT * FROM workspace.sql_course_db.customers
```

#### Step-2: Combining the tables

```sql
SELECT 
    FirstName AS `First Name`,
    LastName AS `Last Name`
FROM workspace.sql_course_db.employees

UNION

SELECT 
    FirstName, 
    LastName
FROM workspace.sql_course_db.customers
```

## 2. UNION ALL OPERATOR

### _Example 2: Combine the data from employees and customers into one table including duplicates._

#### Step-1: Query the employees and customers tables.

```sql
SELECT * FROM workspace.sql_course_db.employees
```

```sql
SELECT * FROM workspace.sql_course_db.customers
```

#### Step-2: Combing the tables including duplicates.

```sql
SELECT 
    FirstName AS `First Name`,
    LastName AS `Last Name`
FROM workspace.sql_course_db.employees

UNION ALL

SELECT 
    FirstName, 
    LastName
FROM workspace.sql_course_db.customers
```

## 3. EXCEPT OPERATOR

### _Example 3: Find employees who are not customers at the same time._

#### Step-1: Query the customers and employees tables.

```sql
SELECT * FROM workspace.sql_course_db.customers
```

```sql
SELECT * FROM workspace.sql_course_db.employees
```

#### Step-2: Finding only the customers who are not employees.

```sql
SELECT 
    FirstName, 
    LastName
FROM workspace.sql_course_db.customers

EXCEPT

SELECT 
    FirstName AS `First Name`,
    LastName AS `Last Name`
FROM workspace.sql_course_db.employees
```

## 4. INTERSECT OPERATOR

### _Example 4: Find employees who are also customers._

```sql
SELECT 
    FirstName AS `First Name`,
    LastName AS `Last Name`
FROM workspace.sql_course_db.employees

INTERSECT

SELECT 
    FirstName, 
    LastName
FROM workspace.sql_course_db.customers
```

# CHALLENGE

### _Example 5: Orders are stored in seperate tables (Orders and OrdersArchive). Combine all orders into one report without duplicates._

#### Step-1: Query of Orders and OrdersArchive Tables.

```sql
SELECT * FROM workspace.sql_course_db.orders
```

```sql
SELECT * FROM workspace.sql_course_db.orders_archive
```

#### Step-2: Combining the Tables using UNION SET Operator.

```sql
SELECT
    'orders' AS SourceTable,
    OrderID,
    ProductID,
    CustomerID,
    SalesPersonID,
    OrderDate,
    ShipDate,
    OrderStatus,
    ShipAddress,
    BillAddress,
    Quantity,
    Sales,
    CreationTime
FROM 
    workspace.sql_course_db.orders

UNION

SELECT 
    'orders_archive' AS SourceTable,
    OrderID,
    ProductID,
    CustomerID,
    SalesPersonID,
    OrderDate,
    ShipDate,
    OrderStatus,
    ShipAddress,
    BillAddress,
    Quantity,
    Sales,
    CreationTime
FROM 
    workspace.sql_course_db.orders_archive
ORDER BY
    OrderID
```
