# 1. BASIC JOINs

## 1.1 NO JOIN

### _Example 1. Retrieve all data from customers and products as seperate results_

```sql
SELECT * FROM workspace.salesdb.dim_customers;
```

```sql
SELECT * FROM workspace.salesdb.dim_products;
```

## 1.2 INNER JOIN

### _Example 2. Get all customers along with their orders,  but only for customers who have placed an order_

```sql
SELECT 
    c.CustomerID,
    c.FirstName,
    o.OrderID
FROM workspace.sql_course_db.customers AS c
INNER JOIN workspace.sql_course_db.orders As o
ON c.CustomerID = o.CustomerID
ORDER BY c.CustomerID ASC
```

## 1.3 LEFT JOIN

### _Example 3. Get all the customers along with their orders including those without orders_

```sql
SELECT
    c.CustomerID,
    c.FirstName,
    o.OrderID
FROM workspace.sql_course_db.customers AS c
LEFT JOIN workspace.sql_course_db.orders AS o
ON c.CustomerID = o.CustomerID 
ORDER BY c.CustomerID ASC
```

# 1.4 RIGHT JOIN

### _Example 4. Get all the customers along with their orders including orders without matching customers (RIGHT JOIN)_

```sql
SELECT
    c.CustomerID,
    c.FirstName,
    o.OrderID
FROM workspace.sql_course_db.customers AS c
RIGHT JOIN workspace.sql_course_db.orders AS o
ON c.CustomerID = o.CustomerID 
ORDER BY c.CustomerID ASC
```

### _Example  5. Get all the customers along with their orders including orders without matching customers (LEFT JOIN)_

```sql
SELECT
    o.OrderID,
    c.FirstName,
    c.CustomerID
FROM workspace.sql_course_db.orders AS o 
LEFT JOIN workspace.sql_course_db.customers AS c
ON c.CustomerID = o.CustomerID 
ORDER BY OrderID ASC
```

## 1.5 FULL JOIN

### _Example 6. Get all the customers and all order even if there is no match._

```sql
SELECT 
    c.customerID,
    c.FirstName,
    o.OrderID
FROM workspace.sql_course_db.customers AS c
FULL JOIN dbacademy.default.orders AS o
ON c.CustomerID = o.CustomerID 
ORDER BY c.CustomerID ASC
```

# 2. ADVANCED JOINs

## 2.1 LEFT ANTI JOIN

### _Example 7: Get all the customers who haven't placed any order_

```sql
SELECT 
    c.CustomerID,
    c.FirstName,
    o.OrderID
FROM workspace.sql_course_db.customers AS c
LEFT JOIN workspace.sql_course_db.orders AS o
ON c.CustomerID = o.CustomerID 
WHERE o.CustomerID IS NULL
ORDER BY c.CustomerID ASC
```

## 2.2 RIGHT ANTI JOIN

### _Example 8: Get all orders without matching customers_

```sql
SELECT 
    o.OrderID,
    c.FirstName,
    c.CustomerID   
FROM workspace.sql_course_db.customers AS c
RIGHT JOIN workspace.sql_course_db.orders AS o
ON c.CustomerID = o.CustomerID 
WHERE c.CustomerID IS NULL
ORDER BY o.OrderID ASC
```

### _Example 9: Get all orders without matching customers (Using LEFT JOIN)_

```sql
SELECT 
    o.OrderID,
    c.FirstName,
    c.CustomerID   
FROM workspace.sql_course_db.orders AS o 
LEFT JOIN workspace.sql_course_db.customers AS c
ON o.CustomerID = c.CustomerID 
WHERE c.CustomerID IS NULL
ORDER BY o.OrderID ASC
```

# 2.3 FULL ANTI JOIN

### _Example 10: Find customers without orders and orders without customers_

```sql
SELECT 
    c.CustomerID,
    c.FirstName,
    o.OrderID
FROM workspace.sql_course_db.customers AS c
FULL JOIN workspace.sql_course_db.orders o 
ON c.CustomerID = o.CustomerID 
WHERE c.CustomerID IS NULL 
OR o.OrderID IS NULL
ORDER BY c.CustomerID ASC
```

# CHALLENGE

### _Example 11: Get only those customers who have placed orders without using INNER JOIN._

```sql
SELECT
    c.CustomerID,
    c.FirstName,
    o.OrderID
FROM
    workspace.sql_course_db.customers AS c
LEFT JOIN
   workspace.sql_course_db.orders AS o 
ON
    c.CustomerID = o.CustomerID
WHERE
    o.OrderID IS NOT NULL
```

## 2.4 CROSS JOIN

### _Example 12: Get all possible combination of customers and orders_

```sql
SELECT 
    *
FROM
    workspace.sql_course_db.customers
CROSS JOIN
    workspace.sql_course_db.orders
```

# CHALLENGE

### _Example 13: Retrieve a list of orders along with the related customers, products, and employee details. For each order, display: Order ID, Customer's name, Product name, Sales, Price, Salesperon's name._

#### Step 1: Analyze customers, orders, products, and employee tables

```sql
-- Query of Orders Table
SELECT * FROM workspace.sql_course_db.orders
```

```sql
-- Query of Products Table
SELECT * FROM workspace.sql_course_db.products
```

```sql
-- Query of Customers Table
SELECT * FROM workspace.sql_course_db.customers
```

```sql
-- Query of Employees Table
SELECT * FROM workspace.sql_course_db.employees
```

#### Step 2: Retrieving the list of orders along with the related customers, products, and employees

```sql
SELECT
    o.orderid AS `Order ID`,
    concat(c.FirstName, ' ', c.LastName) AS `Customer's Name`,
    p.Product AS `Product's Name`,
    o.Sales AS `Sales`,
    p.Price AS `Price`,
    concat(e.FirstName, ' ', e.LastName) AS SalesPerson
FROM
    workspace.sql_course_db.orders AS o
LEFT JOIN 
    workspace.sql_course_db.products AS p
    ON o.ProductID = p.ProductID
LEFT JOIN
    workspace.sql_course_db.customers AS c
    ON o.CustomerID = c.CustomerID
LEFT JOIN
    workspace.sql_course_db.employees AS e
    ON o.SalesPersonID = e.EmployeeID
```
