# 6. Python Numbers

## 6.1 Types

#### _Example 1:  Find out the type of the numbers._

```python
x = 5, # is an interger
y = 5.7, # is a float
z = 2 + 3j, # is a complex number
print(type(x));
print(type(y));
print(type(z));
```

#### _Example 2:  Convert a string to an integer._

```python
# Use case-1: string ==> integer
a = "24"; # is a string
print(type(a)); # showing the type of the variable
print(a * 3); # multiplying the string
a1 = int(a)  # converting the string to an integer
print(type(a1)); # showing the type of the variable
print(a1 * 3); # multiplying the integer

# Use case-2: int <=> float
b = 24.5; # is a float
print(type(b)); # showing the type of the variable
b1 = int (b)  # converting the float to int
print(b1);
print(type(b1)); # showing the type of the variable
b2 = float(b1); # converting float to int
print(b2); # showing the type of the variable
print(type(b2));

# Use case-3: complex number
x1 = 3; # is the real part
x2 = 4; # is the imaginary part
x3 = complex(x1, x2);
print(x3);
print(type(x3));
```

## 6.2 Math Operators

#### _Example 3:  Show the use cases of Math Operators._

```python
print(2+2) # addition
print(2-2) # subtraction
print(2*2) # multiplication
print(2/2) # division
print(2**3) # exponentiation
print(2%2) # modulus
print(2//2) # floor division
```

#### _Example 3:  Show the shorthand assignment._

```python
m = 3
print(m);
m += 2 # adding 2 to m
print(m);
m -= 2 # subtracting 2 from m
print(m);
m *= 2 # multiplying m by 2
print(m);
m /= 2 # dividing m by 2
print(m);
```

## 6.3 Rounding

#### _Example 4:  Show the use cases of Rounding._

```python
import math # importing math module
price = 35.5428

# round vs floor vs ceil
# among round, floor and ceil, round is the most commonly used
# ceil is mainly used in data engineering specially splitting data into pages or bathches
print(round(price, 2)); # rounding to 2 decimal places
print(round(price)); # rounding to the nearest integer
print(math.floor(price)); # this is python module. so, we have to import the module math first. using the floor function rounding it down to the nearest integer
print(math.ceil(price)); # this is python module. so, we have to import the module math first. using the ceil function rounding it up to the nearest integer

# trunc vs int
# if you only have to import math module for trunc, then it is better to int instead of trunc
print(math.trunc(price)); # using trunc to remove all the decimal part. we can do the same by changing the variable type from float to int
print(int(price));

# round half to even
# if we have some numbers like 1.5, 2.5, 3.5, these numbers will always round to 2, 2, 4 respectively.
roundPrice1 = 1.5
roundPrice2 = 2.5
roundPrice3 = 3.5
print(round(roundPrice1));
print(round(roundPrice2));
print(round(roundPrice3));
```

## 6.6 Validation

#### _Example 6:  Show the use cases of Validation._

```python
# Use case-1: Checking whether a number is a whole number or not
s = 7.0
print(s.is_integer()); # whole number because it has no decimal part
t = 7.5
print(t.is_integer()); # not whole number because it has decimal part

# Use case-2: Checking whether a number belongs to a certain data type or not
u = 70
print(isinstance(u, int)); # an integer
print(isinstance(s, int)); # not an integer because it has decimal part
print(isinstance(u, float)); # not a float
print(isinstance(s, float)); # a float
```
