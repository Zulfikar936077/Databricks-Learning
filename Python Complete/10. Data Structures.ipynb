### 10.1.3 Unpack Lists

```python
# Why needed?
# Let's say, ['Maria', 29, 'Data Engineer', 'Berlin'] is a List.
# May be we need to put these values into different variables, like name, age, job, city for further operations.
# So, we can use unpacking to do it in a clean way.

person = ['Maria', 29, 'Data Engineer', 'Berlin']

# Way-1: Regular Indexing. Difficult for longer lists
# name = person[0]
# age = person[1]
# job = person[2]
# city = person[3]


# Way-2: Unpacking
name, age, job, city = person
print(name, age, job, city);
# updating name
person[0] = 'John'
print(person);
# adding item to the List
person.append('Germany') # adding item to the List
name, age, job, city, country = person
print(person[4]);


## Use of Asterisk(*)/ Rest Collector:
### What if I just want the first and last item in the List
name, *details, country = person
print(name); # 1st variable
print(details); # in-between
print(country); # last variable

### What if I just want the 1st variable
name, *details= person
print(name); # 1st variable
print(details); # rest

## Unpacking Rule:
### Nr. of variables must match the values exactly - not mless, not more

## Skipping items (_):
### If I just need specific variables from the List
name, _, job, _ , _= person
print(name, job);

## Asterisk(*)+ Underscore (_):
name, *_ = person
print(name);
print(_);
```

#### _There are usually 3 types of changes of data:_
        - Adding new item
        - Removing bad data
        - Updating data

```python
members = ['a', 'b', 'c']


# 1. Adding data

## a. Adding item at the end of the list
members.append('d')      # printing this line will not show the result, but the next line with the printing the main variable will show it
members.append('e')
print(members)

## b. Adding item at any position of the list
members.insert(0, 1) # here, 0 is the index number and 1 is the value that I want to insert
print(members);

members.insert(6, 'f') # remember, 1 is already added at the 1st position, so the index number is also changed, be careful
print(members);

## c. Adding list in the matrix
matrix = [[1, 2, 3],   # Row 0
          [4, 5, 6],   # Row 1
          [7, 8, 9]]   # Row 2
print(matrix);


matrix.append(['x', 'y', 'z']) # adding a list at the end of the matrix
print(matrix);

matrix.insert(0, ['a', 'b', 'c']) # adding a list at the beginning of the matrix
print(matrix);

matrix[1].append('x') # adding a value at the end of the 2nd row
print(matrix);

matrix[0].insert(0, 'z') # adding a value at the beginning of the 1st row
print(matrix);

```

```python
# 2. Remove items
members1 = ['a', 'b', 'c']

## a. Removing all the items from the list
members1.clear()
print(members1);

## b. Removing the first matching item from the list
members2 = ['a', 'b', 'a']
members2.remove('a')   # removes the 1st matching item from the list
members2.remove('a')   # removes the 2nd matching item from the list
print(members2); 

## c. Removing any item from a specific position
members3 = ['a', 'b', 'c']
members3.pop(0) # removes a from the list
removed= members3.pop()  # if any index number is not given in .pop, it will remove the last item from the list
print(members3);
print("Removed item: ", removed);  # it also shows the removed item

## d. Removing lists and items from matrix
matrix = [[1, 2, 3],   # Row 0
          [4, 5, 6],   # Row 1
          [7, 8, 9]]   # Row 2
print(matrix);

matrix.remove([1, 2, 3]) # removing the list from the matrix
print(matrix)

matrix.pop() # removing last list from the matrix
print(matrix);

matrix[0].remove(5) # removing 5 from the remaining row (now at index 0 after previous removals)
print(matrix);

matrix[0].pop(1) # removing 6 from the matrix (index 1 after previous removals)
print(matrix);

```

```python
# 3. Updata data
members4 = ['a', 'b', 'c']

## a. In case of single item
members4[0] = 'd'  # changing the 1st item of the list
print(members4);

## b. In case of multiple items
matrix = [[1, 2, 3],   # Row 0
          [4, 5, 6],   # Row 1
          [7, 8, 9]]   # Row 2

matrix[0] = [10, 11, 12] # a.changing the 1st row of the matrix
print(matrix);

matrix[0][0] = "-"  # b. changing the 1st item of the 1st row of the matrix. The first 0 is the row nr. and the second 0 is the item nr.
print(matrix); 
```

```python
# Sorting List

alphabets = ['a', 'c', 'b']

# 1. Sorting of the regular list
## a. Sorting in ascending order
alphabets.sort()
print(alphabets);

## b. Sorting descending order
alphabets.sort(reverse = True)
print(alphabets); 

# 2. Sorting of matrix
matrix = [ [7, 8, 9],    # Row 0
           [1, 2, 3],    # Row 1 
           [4, 6, 5],     # Row 2
          ]   

## a. Sorting the matrix
matrix.sort()
print(matrix);

## b. Sorting a specific list of the matrix
matrix[1].sort() # Index 1, because [4,6,5] list changed its index nr. in the last print from 2 to 1
print(matrix);

# 3. Making a copy of the list to leave the original list unchanged
## a. ascending order
sortedList = sorted(alphabets)
print("Sorted List: ", sortedList);

## b. descending order
sortedList = sorted(alphabets, reverse = True)
print("Sorted List: ", sortedList);


# 4. Reversing the list
alphabets1 = ['a', 'c', 'b']

## a. Reversing the list
alphabets1.reverse() # Reverse is diff. method than descending. It just reverses the order of the list, does not descend it. 
print(alphabets1);

## b. Reversing the list and making a copy of the original list
new_List = list(reversed(alphabets1)) # reversed is used to create a copy of the original list like sorted, but it returns objects, so use list() to get the copied 
print("Reversed List: ", new_List);
```

### 10.1.7 Copying Lists

![image_1793356128063.png](./image_1793356128063.png "image_1793356128063.png")

```python
# 1. Assignment =
# Assignment = does not make another copy of the original List, it is used as Reference.
# Attention: Be careful before using Assignment =, as removing one item from ref list can remove the item from the original list

alphabets = ['a', 'c', 'b']
alphabets_copy = alphabets

print("Original List: ", alphabets);
print("Copied List: ", alphabets_copy);

alphabets_copy.append('d')  # adding 'd' to the copied list
print("Original List: ", alphabets); # 'd' is also added to the original list as copied list is the reference list if original list
print("Copied List: ", alphabets_copy);

alphabets.pop() # removing last item from the original list
print("Original List: ", alphabets); # the last item is also removed from the copied list as copied list is the reference list if original list
print("Copied List: ", alphabets_copy);
```

```python
# 2. Shallow copy
# It will create an independent copy of the original list, so removing an item from the copied list will not remove the item from the original list
# Safer method compared to Assignment =

alphabets = ['a', 'c', 'b']
alphabets_copy = alphabets.copy() # creating an independent copy of the list    #copy.copy() is more general than list.copy()
print("Original List: ", alphabets);
print("Copied List: ", alphabets_copy);

alphabets_copy.append('d')  # adding 'd' to the copied list
print("Original List: ", alphabets); # 'd' is not added to the original list
print("Copied List: ", alphabets_copy); # 'd' is added to the list

alphabets.pop() # removing last item from the original list
print("Original List: ", alphabets); # the last item is not removed from the copied list
print("Copied List: ", alphabets_copy); # the last item is not removed from the copied list
```

![image_1793354234020.png](./image_1793354234020.png "image_1793354234020.png")

```python
# 3. Deep Copy
# In copy only the Top_Level (List Levels/ list indicating lists) are Independent (lists pointing to the items), not the items inside the sub-list (Deeper Level).
# That's why it is called shallow copy
# For example:

matrix = [['a', 'b'],
          ['c', 'd'],
          ['e', 'f']]

# step-1: creating an independent copy of the list
matrix_copy = matrix.copy() 
print("Original List: ", matrix);
print("Copied List: ", matrix_copy);

# step-2: changes are independent in the top level
matrix.pop() # removing the last list from the copied list
print("Original List: ", matrix); # the last list is removed from the original list
print("Copied List: ", matrix_copy); # the last list is not removed from the copied list

# step-2: changes are not independent in the deep level
matrix_copy[0].append('z') # adding an item 
print("Original List: ", matrix); # the item is added in both the lists
print("Copied List: ", matrix_copy); # 

```

![image_1793355092047.png](./image_1793355092047.png "image_1793355092047.png")

```python
# The problem of shallow copy can be solved by using deepcopy
# In deepcopy, the top level and the deep level are independent, so removing an item from the copied list will not remove the item from the original list
# This is the safest way of copying lists
# If we have to deal with List instead of matrix, we can use copy.copy to make the shallow copy

import copy # importing copy module
matrix = [['a', 'b'],
          ['c', 'd'],
          ['e', 'f']]

# step-1: creating an independent copy of the list
matrix_copy = copy.deepcopy(matrix) 
print("Original List: ", matrix);
print("Copied List: ", matrix_copy);

# step-2: changes are independent in the top level
matrix.pop() # removing the last list from the copied list
print("Original List: ", matrix); # the last list is removed from the original list
print("Copied List: ", matrix_copy); # the last list is not removed from the copied list

# step-3: changes are independent in the deep level
matrix_copy[0].append('z') # adding an item 
print("Original List: ", matrix); # the item is not added in the original list
print("Copied List: ", matrix_copy); # the item is added in the copied list
```

```python
# 4. IS Operator Testing

original = [
    ['a', 'b'],
    ['c', 'd'],
    ['e', 'f']
]

# a. Testing Assignment =
copy1 = original
print("Same Object?", original is copy1); # Same object
print("Shared List?", original[0] is copy1[0]); # Shared list

# b. Testing Shallow Copy
copy2 = original.copy()
print("Same Object?", original is copy2); # Different object
print("Shared List?", original[0] is copy2[0]); # Shared list

# c. Testing Deep Copy
copy3 = copy.deepcopy(original)
print("Same Object?", original is copy3); # Different object
print("Shared List?", original[0] is copy3[0]); # Different List

```

```python
# 1. Transformation of items
letters = ['a', 'b', 'c']
new_letters =[]  # declaring the empty list where the tranformed data will be saved
for l in letters:
    new_letters.append(l.upper())
    print(new_letters);
```

```python
# 2. Iterators: enumorate/ reversed/ zip

## a. enumerate()
        ### enumerate() returns both value and index of an item.
        ### it generally starts at index 0, but one can set the required index.
letters = ['a', 'b', 'c']
print(list(enumerate(letters))); # list() is used to convert the iterator to a list, generally index starts at 0
print(list(enumerate(letters, start=1))); # list() is used to convert the iterator to a list, this code can be used if we want to start the index from 1 or other number

## But we do not use enumerate() as above
letters = ['a', 'b', 'c']
for index, value in enumerate(letters):  # here, we have 2 variable, index and value are the names of the variables
    print(index, value);

## b. reversed()
letters = ['a', 'b', 'c']
for l in reversed(letters):  # reversed() is used to flip the data
    print(l);

## c. zip()
        ### zip() is used to create a list of tuples pairing items of both the lists
letters = ['a', 'b', 'c']
numbers = [1, 2, 3]
for l, n in zip(letters, numbers):
    print(l, n);
```

### 10.1.10 Transforming Lists (Advanced Data Structures)

```python
# d. map()
        ### map(Function, Iterable) # Function that I want to use and Iterable is the List that I want to go through the Function.
        ### map() is fast and clean way to do data transformation.

## Task-1: Convert the list of strings to uppercase
letters = ['a', 'b', 'c']
print(list(map(str.upper, letters))); # use this line if you want to see the output in a list
for l in map(str.upper, letters): # use this line if you want to put the output in a for loop
    print(l);
    
## Task-2: Convert each item of the list to int
numbers = ['1', '2', '3']
print(list(map(int, numbers))); # use this line if you want to see the output in a list
for n in map(int, numbers): # use this line if you want to put the output in a for loop
    print(n);

## Task-3: Clean up the list by removing all unwanted spaces
names = ['  John', '  Mary', '  Bob']
print(list(map(str.strip, names)));  # use this line if you want to see the output in a list
for n in map(str.strip, names): # use this line if you want to put the output in a for loop
    print(n);
```

```python
# Basic Concept of Lambda:
        # Python has many built-in Funtions.
        # However, sometimes we need quick and custom logics.
        # Lambda can be used to serve that need.
        # The Lambda Function is any Function without a name.
        # That's why it is also called anonymous Function.
        # Instead of writing a whole Function, Lambda can be a short way of refering that specific function.
        # Lambda is generally used a lot with other Functions such as map(), filter(), sorted() to perform extra complicated logic on the top of the data structure. 
# How does Lambda works?
        # For Lambda we need two things: a. input variable, b. expression (what will happend with that input variable)
        # Structure: 
                    # Output = lambda X : Expression
        # Example: 
                    # Output = lambda x : x + 1
```

```python
# 1. lambda:    
## Task-1: multiplication
multiple = lambda x: x * 2  # the value of x is passed when the lambda is called in the next line
print(multiple(2));

## Task-2: addition
add = lambda x, y: x + y
print(add(2, 3));

    ## Task-3: checking a value
check = lambda i: i in "Python"
print(check("n"));

# 2. lambda + map()
    ## Three steps to transform: 1. Think about the expression, 2. Put it in Lambda, 3. Put it in map()

## Task-1: convert the following string list into a float list:
prices = ['$12.50', '$9.99', '$100.00']

    # step-1: removing the $ with an empty string
                    ## p.replace('$', '')          # However, the list is still a string list
    # step-2: converting the string list to float list
                    ## float(p.replace('$', ''))
    # step-3: converting into lambda function
                    ## lambda p: float(p.replace('$', ''))
    # step -4: placing the labda function into map()
                    ## map(lambda p: float(p.replace('$', '')), prices)   # however, it will give the object as the output
    # step-5: converting the object into a list
                    ## list(map(lambda p: float(p.replace('$', '')), prices))
    # step- 6: printing the final output
print(list(map(lambda p: float(p.replace('$', '')), prices))); # this is the final output


# 2. lambda + filter()

## Task-1: filter the list of costs to keep only the prices equal or above 100
costs = [120, 30, 300, 80]
print(list(filter(lambda x : x >= 100, costs)));

## Task-2: keep only students with scores higher than 70
students = [['Maria', 85],
            ['Kumar', 90],
            ['Max', 60]]
print(list(filter(lambda row: row[1] > 70, students))); # we know about [row, index]
                                                        # row: because we are checking every row
                                                        # [1]: because index 1 of every row is the score
                                                        # > 70: because we want to keep only the scores higher than 70
# 3. lambda + sorted()

## Task-1: keep only students with names starting with 'M'
students = [['Maria', 85],
            ['Kumar', 90],
            ['Max', 60]]
print(list(filter(lambda row: row[0].startswith('M'), students)));
```

#### _B. List Comprehension_

 - List comprehension is a concise, single-line syntax in Python used to create a new list from an existing iterable (like a list, tuple, or range). 
 - It replaces traditional multi-line for loops, making your code cleaner, more readable, and often faster to execute. 
 - syntax:
`            new_list = [
                expression 
                for item in iterable
                if condition
            ]`

```python
# Task-1: Normalize the domains into standard format
domains = [
    'www.google,com',
    'openai.com',
    'localhost',
    'WWW.DATAWITHAI.COM'
]


cleaned_domain = [
    d.lower().replace('www.', '')              # expression                 # 1. transformation
    for d in domains                           # item in iterable           # 2. iterable
    if '.' in d                                # if condition               # 3. filter
]
print(cleaned_domain);
```

## 10.2 Data Behaviour (Tuples, Set, and Dictionary)

### Charactarictics of Data Types:
There are main four characteristics of Data Types:
- Ordered (printing in the same sequence it is given in the data structure)
- Duplicates (more than one similar value) 
- Indexed (positioning in the data structure)
- Mutables (insert, remove, update, sort)

### 10.2.1 Data Structure (Tuple)

```python
# Tuple is an ordered collection that can be changed after creating it
# So, Tuple is immutable
# Use case: each time we feel that the data must be protected, we should use tuple

my_tuple = (1, 3, 2, 3)
print(my_tuple);   # ordered, duplicates
print(my_tuple[1]); # indexed
print(sorted(my_tuple)); # sorted but the output will be a list

# my_tuple[1] = 4; # immutable/ unchangable
# my_tuple.remove() # immutable
# my_tuple.pop() # immutable
# my_tuple.add() # immutable
```
