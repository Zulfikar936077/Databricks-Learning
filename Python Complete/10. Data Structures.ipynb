### 10.1.1 Create Lists

#### _10.1.1.1 One Dimensional/ One Row Lists_

```python
# when we create a List, it creates an object behind the scene, inside this object there will be two sections, one is the data section type and the other is the pointer. For each value in the List individual sub-object will be created and the pointer will be pointing to the sub-objects. The array is the collection of these sub-objects.

# Creating List
empty = []              # creating an empty List
print(empty);           # displaying the List
print(type(empty));     # displaying the type of List

# Way-1: Entering the value in the List manually
letters = ['a', 'b', 'c']
print(letters);
print(type(letters));

# Way-2: converting an iterable(sequence) into a list
names = list('Python')
print(names);
numbers = list(range(5))
print(numbers)
```

#### _10.1.1.2 Two Dimensional/ Nested Lists_

```python
# In case of Nested Lists, there will two rows of list. 
# It will create matrix as the pointer from the object will be pointing to the objects of each row, then the pointers of sub-objects will be pointing to the values of the sub-objects.

matrix = [[1, 2, 3], 
          [4, 5, 6], 
          [7, 8, 9]]
print(matrix)
print(type(matrix))

mixed_matrix = [[1, 2, 3],
                ['a', 'b'],
                [True]]
print(mixed_matrix);
print(type(mixed_matrix))
```

### 10.1.2 Access & Read Lists

```python
# access & read lists means indexing and slicing

# 1. Indexing:
## Way-1: aceessing & reading a 1 dimensional list
st = ['a', 'b', 'c']
print(st);
print(st[0]);
print(st[-1]);

## Way-2: accessing & reading a 2 dimensional list
# here, row & column/index number must be indicated
matrix = [[1, 2, 3],   # Row 0
          [4, 5, 6],   # Row 1
          [7, 8, 9]]   # Row 2
print(matrix);      # selecting whole list
print(matrix[1][1]); # selecting specific item
print(matrix[-1]); # selecting the whole row

# 2. SLicing:
# Index is used when we have to extract one item from the List, whereas Slicing is used to extract multiple items from the List.
print(st[0:2]); # slicing first 2 char
print(st[:2]); # same as the 1st one
print(st[1:3]); # slicing 2nd & 3rd char
print(st[1:]); # same as the 2nd one
print(matrix[0:2]) # slicing first 2 rows
print(matrix[:2]) # smae as the 3rd one
print(matrix[1:3]) # slicing 2nd & 3rd rows
print(matrix[1:]) # same as the 4th one
print(matrix[-1][0:2]); # slicing from specific row

```

### 10.1.3 Unpack Lists

```python
# Why needed?
# Let's say, ['Maria', 29, 'Data Engineer', 'Berlin'] is a List.
# May be we need to put these values into different variables, like name, age, job, city for further operations.
# So, we can use unpacking to do it in a clean way.

person = ['Maria', 29, 'Data Engineer', 'Berlin']

# Way-1: Regular Indexing. Difficult for longer lists
# name = person[0]
# age = person[1]
# job = person[2]
# city = person[3]


# Way-2: Unpacking
name, age, job, city = person
print(name, age, job, city);
# updating name
person[0] = 'John'
print(person);
# adding item to the List
person.append('Germany') # adding item to the List
name, age, job, city, country = person
print(person[4]);


## Use of Asterisk(*)/ Rest Collector:
### What if I just want the first and last item in the List
name, *details, country = person
print(name); # 1st variable
print(details); # in-between
print(country); # last variable

### What if I just want the 1st variable
name, *details= person
print(name); # 1st variable
print(details); # rest

## Unpacking Rule:
### Nr. of variables must match the values exactly - not mless, not more

## Skipping items (_):
### If I just need specific variables from the List
name, _, job, _ , _= person
print(name, job);

## Asterisk(*)+ Underscore (_):
name, *_ = person
print(name);
print(_);
```
