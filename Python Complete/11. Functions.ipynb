## 11.3 Types of Functions?

#### 1. Built-in Functions:
- That we use in Python by default.
- Example: print(), len(), str(), input(), type(), bool(), etc.

#### 2. Standard Library:
- Written by Python Team.
- But we have to import it to use it.
- Example: math, datetime, random, etc.

#### 3. External Library:
- Written by community members.
- Therefore, we need to install, then import, and then we can use it.
- Example:
  i. Pandas (used for data cleaning, analyzing, exploring, and manipulation)
  ii. Matplotlib (used as low level graph plotting in python that serves as a visualization tool)
  iii. NumPy (used for working with arrays)

#### 4. User-Defined Functions:
- In some cases, the problems are very specific and cannot be solved by these libraries.
- In that case we have to define and use our own Funcions.
