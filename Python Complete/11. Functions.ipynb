## 11.7 Parameters, Local Variables & Global Variables

```python
# Global Variable: created outside the function and can be accessed anywhere
# Local Variable: created inside the function and can be accessed only inside the function
# There are two important questions we need to ask when handling parameters, local variables, and global variables:
            # 1. How long does it live?
            # 2. Where is it accessible?
```

![image_1794819901086.png](./image_1794819901086.png "image_1794819901086.png")

```python
f = 2                   # f is the global variable that can be used inside any function
def multile_factor(x):  # x is the parameter
    y = x * f           # f is used instead of numbers that we used before, y is the local variable
    print(y)            # the result of y is the final result
multile_factor(2)       # 2 is the argument and the value of x
```

![image_1794820346437.png](./image_1794820346437.png "image_1794820346437.png")

```python
# Use case-1: Use of Parameter and Local variable in cleaning data
def clean__text(text):                  # text is the parameter
    cleaned = text.strip().capitalize() # cleaned is the local variable
    print("Raw Text: ", text)
    print("Cleaned Text: ", cleaned)
clean__text(" mArCo  ")
```

```python
# Use case-2: Using Global Variable in cleaning data
case_rule = "lower"                     # case_rule is the Global Variable which is saying the case rule
def clean_text(text):                   # text is the parameter
    cleaned = text.strip()              # cleaned is the local variable
    if case_rule == "lower":            # if the case_rule is lower,
        cleaned = cleaned.lower()       # the text will be lower case
    elif case_rule == "upper":          # if the case_rule is upper,
        cleaned = cleaned.upper()       # the text will be upper case
    else:                               # if the case_rule is not lower or upper,
        cleaned = cleaned.capitalize()  # the text will be capitalized
    print("Cleaned Text: ", cleaned)
clean_text(" mArCo  ")                  # mArCo is the argument
print(case_rule)                        # Global variable can be called outside the function
```

## 11.8 Positional & Keyword Arguments

```python
# In Python, there are two ways to send to values to the Functions:
            # 1. Positional Arguments
            # 2. Keyword Arguments
# Positional Arguments: values pass to the function based on their order
# Keyword Arguments: values pass to the function based on their name
```

```python
# Use case-1: Building a Full Clean Name
        # 1. Clean the First and Last Names
        # 2. Merge them into Full Name
def clean_name(first_name, last_name, country):
    first = first_name.strip().capitalize()
    last = last_name.strip().capitalize()
    full = first + " " + last
    print("Full Name: ", full)
    print("The person is from", country)
clean_name(" mAria  ", " RODRIGUEZ", "Germany")                                    # Positionl Argument
clean_name(first_name = " mAria  ", last_name = " RODRIGUEZ", country = "Germany") # Keyword Argument


## Mixed Argument: Always start with Positional Arguments and then Keyword Arguments
def clean_name(first_name, last_name, country):
    first = first_name.strip().capitalize()
    last = last_name.strip().capitalize()
    full = first + " " + last
    print("Full Name: ", full)
    print("The person is from", country)
clean_name(" mAria  ", last_name = " RODRIGUEZ", country = "Germany")               # Mixed Argument
```

## 11.9 Default Parameters

```python
# Dafault Parameter: Default Parameter has already a value so if you don't pass any value, it will use the default value

# Example: Default Parameter
def clean_name(first_name, last_name, country = "Germany"): # country is the default parameter
    first = first_name.strip().capitalize()
    last = last_name.strip().capitalize()
    full = first + " " + last
    print("Full Name: ", full)
    print("The person is from", country)
clean_name(" mAria  ", " RODRIGUEZ") 
# Even though country is not used but it will use the default value which is Germany                      
clean_name(first_name = " mAria  ", last_name = " RODRIGUEZ")  
```

## 11.10 *args & **kwargs

![image_1794827714921.png](./image_1794827714921.png "image_1794827714921.png")

```python
# S/times we don't know how many arguments we will have, so we can use *args and **kwargs
# *args & **kwargs allow function to accept a unknown number of arguments
# *args: positional arguments, should be used when we have similar type of info such as int, str, etc.
# **kwargs: keyword arguments, should be used when we have different types of info 


# Example-1: *args
# If we have increaing arguments
def total(a = 0, b = 0, c = 0, d = 0):  # Here, every time we need to add the parameter & the nr. of arguments is unknown
    print(a + b + c + d)
total(1, 2, 3, 4)

# Therfore, we can use *args
def total(*args):          # * is used only in the parameter
    print(type(args))      # the type of *args is tuple, so we can use any method 
    print(sum(args))
total(1,2,3)

# Example-2: **kwargs
def create_user(**kwargs):
    print(type(kwargs))     # the type of **kwargs is dictionary because we are using keyword arguments
    print(kwargs)
create_user(first = "Maria", last = "Rodriguez", age = 30, country = "Germany")
```

## 11.11 Return

![image_1794829410084.png](./image_1794829410084.png "image_1794829410084.png")

```python
# return replaces the print inside the function
# then the return is stored in the variable
```

```python
# Example-1: With a name and Single Return

def clean_name(name):                   # name is the parameter
    cleaned = name.strip().capitalize() # cleaned is the local variable 
    return cleaned                      # return is the output, used instead of print()
cln_name = clean_name(" mAria  ")       # cln_name is the global variable
print(cln_name)                         # printing the global variable

# Example-2: Multiple names and Multiple Return Options
def clean_name(name):
    if not name:
        return None                         # return nr. 1
    else:
        cleaned = name.strip().capitalize()
        return cleaned                      # return nr. 2
cln_name = clean_name("")                   # will return None
print(cln_name)

# Example-3: Single Name and Multiple Returns
def clean_name(name):
        cleaned_cap = name.strip().capitalize()
        cleaned_lo = name.strip().lower()
        cleaned_up = name.strip().upper()
        return cleaned_cap, cleaned_lo, cleaned_up      # Multiple return of the same name
cln_name = clean_name(" mAria  ")          
print(cln_name)                                         # the value will be in tuple


# we can unpack the tuple
clean_cap,  clean_lo, clean_up = clean_name(" mAria  ") 
print(clean_cap)
print(clean_lo) 
print(clean_up)
```

```python
# Task-1: Stores application log messages in a file whenever an event occurs.
def write_log(message):
    with open(r"C:\Practice\app.log", "a") as file:   # r is for raw string     # a is the method append
        file.write(message + "\n")
write_log("Application started")
```

#### _11.12.2 Transformation Function_

```python
# Task-1: Clean email addresses and split them into username and domain.
def clean_email(email):
    email_cln = email.strip().lower()
    username, domain = email_cln.split("@")
    return {"Username: ": username, 
            "Domain ":domain}
clean_email("  Maria.Rodriguez@company.com  ")
```

```python
# Task-2: Check whether an email has a basic valid format.
def valid_email(email):
    return "@" in email and "." in email
valid_email("maria.rodriguez@company.com")
```

#### _11.12.4 Orchestrator Function_

```python
# Orchestrator function controls control flow by calling other functions in the correct order.
```

### Project
- Receive an email from the user
- Validate the email
- If it is invalid, log an error in a file
- If it is valid, Clean and structure the email
- Log each step of the program

```python
# There are two ways to do the project:

# Way-1:
write_log("Application started")                # 5. Writing log           
email = input("Please enter your email: ")      # 1. Receiving the user's email

if not valid_email(email):                      # 3. Validating the email 
    write_log("Invalid email address")
else:
    clean_email(email)                          # 4. Cleaning the email
    write_log("Email address is valid")
write_log("Application finished")               # 5. Writing log


# Way-2: Using Orchestrator Function

## Step-1: Orchestrator Function
def process_user_email(email):
    write_log("Application started")                                              
    
    if not valid_email(email):                     
        write_log(f"Invalid email address: {email}")
    else:
        clean_email(email)                        
        write_log(f"Email address is valid: {clean_email}")

    write_log("Application finished") 

# Step-2: Receiving the email from the user
email = input("Please enter your email: ") 

# Step-3: Calling the orchestrator function
process_user_email(email)
```

## 11.13 Writing Clean Functions

#### The professional rules to follow:
1. Use snake casing for function name.
2. Show the data types of the paramaters and expected data type.
3. Use doc string and explain the purpose of the function.
4. Describe about the arguments and returns.
5. Always use local variables with different name than the paramater to save the results.

```python
# Follow PEP8 - Style Guide
```

```python
# Bad Function

def DiscPrint( p, r):
    print("calculating discount")
    p = p -(p* r/100)
    print(p)
```

```python
# clean function
def discount_price(price: float, rate: float) -> float:             # Rule: 1, 2, 
    """                                                             # Rule: 3
    Calculate the final price after applying a discount.            # Rule: 4
    Args:
        price(float): The original price.
        rate(float): The discount rate in percentage.
    Returns:
        float: The final price after discount.
    """
    final_price = price - (price * rate / 100)                      # Rule: 5
    return final_price
```
