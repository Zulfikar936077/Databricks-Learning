```python
# Use cases: Standalone if, elif, else, Nested if

score = 90;
submitted_project = True;

if score >= 90:
    if submitted_project:    # Here, we don't have to write submitted_project == True, submitted_project: means it is True
        print("A+");
    else:
        print("A");
elif score >= 80:
    print("B");
elif score >= 70:
    print("C");
elif score >= 60:
    print("D");
else:
    print("F");
```

```python
# Use cases: Connecting Conditions

score = 90;
submitted_project = True;

if score >= 90 and submitted_project == True:   # "and" operator is used to connect the conditions
    print("A+"); 
elif score >= 90:                               # the previous else became elif here
    print("A");
elif score >= 80:
    print("B");
elif score >= 70:
    print("C");
elif score >= 60 or submitted_project:          # another connecting operator
    print("D");
else:
    print("F");
```

```python
score = 70;
submitted_project = False;

if score >= 90:
    print("High Score.");
else:
    print("Low Score.");

if submitted_project:              # submitted_project: means submitted_project == True
    print("Project submitted.");
else:
    print("Project not submitted.");
```

```python
print("A" if score >= 90 else "B" if score >= 80 else "F"); # it can be used for simple logic
                                                            # but for complex logic it is better to use classical if
```

```python
# match case can be used instead of if-else conditions when the logic is simple

country = "Italy";
match country:
    case "Italy":
        print("IT");
    case "United States" | "USA":             # use | when there is a possibility of double match
        print("US")
    case _:
        print("Unknown Country");              # use " _" when the macthing value is unknown
```

#### _1. Validate the Quality and Correctness of Email Values._
- Must not be empty
- Must contain '.' and '@'
- Must contain exactly one '@' symbol
- Must end with '.com', '.org', or '.net'
- Must not be longer than 254 characters
- Must start and end with a letter or digit

```python
user_email = "unknown.email@gmail.com";
if (user_email != ""):
    print("Email address is not empty.");
if ('.' in user_email and '@' in user_email):
    print("Email address contain '.' and '@'.");
if (user_email.count("@") == 1):
    print("Email contains exactly one '@'.")
if (len(user_email) < 254):
    print("The Email address is not longer than 254 characters.");
if(((user_email[0].isalpha) and (user_email[-1].isalpha)) or ((user_email[0].isnumeric) and (user_email[-1].isnumeric))):
    print("The email address starts and ends with letter or digit.");
else:
    print("The email address does not start and ends with letter or digit.")
```

- Must not be empty
