## 7.1 Boolean Expressions

### 7.1.1 Values (True, False)

```python
print(True) # printing the value
print(False) # printing the value
print(type(True)) # type of the value
print(bool(123)) # converting the value to boolean, returns True as there is value.
print(bool("abc")) # converting the value to boolean, returns True as there is value.

# There is a diff. bet. 0, " ", and None. None means that there is no value at all. The other two (0 and " ") are values, but they are empty values.)
print(bool(0)) # converting the value to boolean, , returns False as there is no value.
print(bool("")) # converting the value to boolean, , returns False as there is no value.
print(bool(None)) # converting the value to boolean, , returns False as there is no value.
```

### 7.1.1 Functions (any, all)

```python
# Use case-1: any and all
email = "" # if space is there, it will be not considered as empty value
phone = "0176-123456"
username = "" # if space is there, it will be not considered as empty value

# use of any() function
if any([email, phone, username]):
    print("User has provided at least one of the contact methods.")
else:
    print("All values are empty.")

# use of all() function
if all([email, phone, username]):
    print("User has provided all contact methods.")
else:
    print("User has not provided all contact methods.")


# Use case-2: isinstances()
print(isinstance(123, int)); # checking whether the value is of type int or not
print(isinstance(True, str)); # checking whether the value is of type str or not

# Use case-3: endswith() and startswith()
print("Hello".endswith("o")) # checking whether the value ends with "o" or not
print("Hello".startswith("H")) # checking whether the value starts with "H" or not
```

## 7.2 Operators

### 7.2.1 Membership Operators (in | not in)

```python
print('o' in "python");
print(2 not in [1,2,3]);
```

#### _Example 1: Validate that the domain is not in the banned list._

```python
domain = "gmail.com";
banned_domain = ["spam.com", "fake.com", "bot.net"];
if (domain in banned_domain):
    print("Your domain is banned.")
else:
    print("Your domain is not banned.");
```

### 7.2.2 Comparison Operators (==, !=, <, >, <=, >=)

```python
# Comparing numbers
print(9 == 9)
print( 9 != 10);
print( 9 > 3);
print(9 >= 3);
print(3 < 9);
print(3 <= 9);

# Comparing strings
print("a" < "b"); # consideration will be done alphabetically
print("a" > "b");
print("a" == "A"); # python is also case sensative

# chained comparison
print(1 < 4 < 8); # chained comparison is done from left to right
print(1 > 4 < 8);
```

#### _Example 2: Is age between 18 and 30?_

```python
age = 20;
print(18 < age < 30);
```

### 7.2.3 Logical Operators (and | or | not)

```python
# in case of "and" operator both the conditions must be true
print(3 < 5 and 5 == 5);
print(3 > 5 and 5 == 5);

# in case of "or" operator at least one of the conditions must be true
print(3 < 5 or 5 == 5);
print(3 > 5 or 3 == 5);
```

#### _Example 3: Is the system under pressure?_

```python
cpu_usage = 70;
memory_usage = 95;
if (cpu_usage > 90 or memory_usage > 90):
    print("The system is under pressure");
else:
    print("Everything is fine.");
```

```python
# "not" operator has priority over "or" operator
print(5 == 5 or 8 > 5 and 6 < 4); # the answer would have been False if the execution started from left hand side

# parenthesis has priority over "and" operator
print((5 == 5 or 8 > 5) and 6 < 4);
```

```python
# use cases of "not" operator

print(not True); # the answer will be False
print(not False); # the answer will be True
print(not not False); # the answer will be False
```

#### _Example 4: Allow access only if the user is logged in or they are guest but they must not be banned?_

```python
is_logged_in = True;
is_guest = False;
is_banned = False;
print(is_logged_in or is_guest and not is_banned);
```

### 7.2.4 Identity Operators (is | is not)

#### _Example 5: Checks if two variables refer to the same object in memory._

```python
# Case-1:
x = [1, 2, 3];
y = [1, 2, 3];
print( x == y); #the comparison operator will compare the value of the variable, that's why it is True
print(x is y); # the identity operator will check the ID of the objects which are diff. for the variable, that's why it is False

# Case-2:
x = 10;
y = 10;
print( x == y); #the comparison operator will compare the value of the variable, that's why it is True
print(x is y); # the identity operator will check the ID of the objects which is same because of the simple value of the variable, that's why it is True

# Case-3:
# If we want to say that the 2nd variable has the same value as the 1st varriable, then to avoid repetion we can do:
x = [1, 2, 3];
y = x;
print(x == y);
print(x is y); # In this case, both the variables are referring to the same objects, so the answer should be True
```

#### _Example 6: Validate the email address. It must be filled in and not empty._

```python
email = None;
print(email is not None and email != ""); # None means unknown but not empty, that's why the 1st condition is False and the 2nd is True 
```
