## 9.1 For Loop

```python
for i in (1,2,3,4,5):
    print(f"Round: {i}");
```

```python
items = (1,2,3,4,5)
for item in items:
    print(f"Round: {item}");
```

```python
# using for loop for letters
letters = " Python"
for letter in letters:
    print(f"Letter: {letter}");
```

```python
# Case-1: using only the stop number of the range
# using for loop using range function for numbers
for number in range(5):                         
    print(f"Number: {number}");

# Case-2: using the start and stop number
for number in range(1,5):
    print(f"Number: {number}");            # number of step is 1

# Case-3: using only the steps
# the default step number is 1
for number in range(1,8,2):                 # number of step is 2
    print(f"Number: {number}");
```

```python
# Real world application of For Loop-1:

scores = [50, 60, 70, 80]
total = 0
for score in scores:
    total += score
    print("Current total:", total)
print("Final total:", total);
```

```python
# Using For Loop to transform data like cleaning data before processing:
# Remember: 1. Clean the data first
#           2. Then transform the data

files = [' Report.csv', 'DATA.csv', ' final.TXT']
for file in files:
    # strip to clean, lower to lowercase, replace to change file type
    file = file.strip().lower().replace('.txt', '.csv')
    print(f"Processing {file}");
```
