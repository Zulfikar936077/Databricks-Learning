## Example 1: Print Statements

```python
print("This is my first python statement.") #single line statement
```

```python
print("This is a multiple line statement. \nThis is the second line.")
```

## Example 2: Single-quote, double-quoute, backslash, mew-line, new-tab, and empty-line

```python
print("Hi, \'Python\'!")  #single-quote
print("Hi, \"Python\"!") #double-quote
print("Hi, \\Python\\!")  #backslash
print("Hi, \nPython!")   #newline
print("Hi, \tPython!")   #tab
```

```python
print("line 1\n") #empty line
print("line 2")
print("line 3\n") #empty line
print("line 4")
```
