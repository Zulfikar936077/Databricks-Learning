# 1. Import Libraries from PySpark

```python
import pyspark.sql.functions as F
from pyspark.sql.window import Window
```

# 2. Read from Silver Table(s) 

# +

# Business Transformation and Modeling

```python
query = """
SELECT
    ROW_NUMBER() OVER (ORDER BY ci.customer_id) AS customer_key,
    ci.customer_id,
    ci.first_name,
    ci.last_name
FROM workspace.silver.crm_customers ci
"""
df = spark.sql(query)
```

# 3. Write it to Gold Table

```python
(
    df
    .write
    .mode("overwrite")
    .format("delta")
    .saveAsTable("gold.dim_customers")
)
```

4. Query the Table

```python
%sql
SELECT * FROM workspace.gold.dim_customers
```
